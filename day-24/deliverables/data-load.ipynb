## Missing Values Audit

```python
def missing_value_audit(data):
    result = pd.DataFrame({
        "column": data.columns,
        "missing_count": data.isna().sum().values,
        "missing_percentage": (data.isna().mean().values * 100)
    })

    return result.sort_values("missing_percentage", ascending=False)
```

## Validation Framework

```python
import pandas as pd
import numpy as np


def validate_rules(df):
    """Execute each rule, return list of dicts for DataFrame construction."""

    n_total = len(df)

    # Pre-compute common masks for efficiency
    mask_cust_null = df["Customer ID"].isna()
    mask_dup_exact = df.duplicated(keep=False)  # marks ALL copies of dup groups
    mask_qty_neg = df["Quantity"] < 0
    mask_price_neg = df["Price"] < 0
    mask_price_zero = df["Price"] == 0
    mask_date_invalid = df["InvoiceDate"].isna()  # assumes already attempted parse
    mask_desc_missing = df["Description"].isna()

    rules = [
        dict(
            rule_id="DQ01",
            category="Missing",
            description="CustomerID should be present",
            columns="Customer ID",
            count=int(mask_cust_null.sum()),
            pct=f"{mask_cust_null.mean()*100:.2f}%",
            severity="Medium",
            review_action="Exclude from customer-scoped analyses (RFM, cohorts); retain in aggregate revenue. Document ~23% anonymity.",
        ),
        dict(
            rule_id="DQ02",
            category="Duplicate",
            description="Exact duplicate rows",
            columns="All",
            count=int(mask_dup_exact.sum()),
            pct=f"{mask_dup_exact.mean()*100:.2f}%",
            severity="Medium",
            review_action="Inspect samples. Drop true repeats (all fields identical incl. timestamp). Keep conflicting records flagged for manual reconciliation.",
        ),
        dict(
            rule_id="DQ03",
            category="Range",
            description="Quantity < 0 (cancellations/returns)",
            columns="Quantity",
            count=int(mask_qty_neg.sum()),
            pct=f"{mask_qty_neg.mean()*100:.2f}%",
            severity="High",
            review_action="Route to dedicated returns table/view. Never mix with positive sales in raw sums. Correlate with Invoice prefix 'C'.",
        ),
        dict(
            rule_id="DQ04",
            category="Range",
            description="Price < 0 (anomalous)",
            columns="Price",
            count=int(mask_price_neg.sum()),
            pct=f"{mask_price_neg.mean()*100:.4f}%",
            severity="High",
            review_action="Investigate each case manually. Determine if double-negative artifact or genuine error. Correct at source or exclude with logged rationale.",
        ),
        dict(
            rule_id="DQ05",
            category="Range",
            description="Price = 0 (promotional/freebie)",
            columns="Price",
            count=int(mask_price_zero.sum()),
            pct=f"{mask_price_zero.mean()*100:.2f}%",
            severity="Review",
            review_action="Flag as non-revenue transactions. Exclude from AOV/margin calcs unless business defines them as valid SKUs. Consider separate promo tracking.",
        ),
        dict(
            rule_id="DQ06",
            category="Validity",
            description="Invalid/unparseable InvoiceDate",
            columns="InvoiceDate",
            count=int(mask_date_invalid.sum()),
            pct=f"{mask_date_invalid.mean()*100:.2f}%",
            severity="CRITICAL",
            review_action="BLOCKING ISSUE IF >0. Fix date parsing format string NOW. Cannot proceed with any temporal analysis until resolved.",
        ),
        dict(
            rule_id="DQ07",
            category="Missing",
            description="Product Description missing",
            columns="Description",
            count=int(mask_desc_missing.sum()),
            pct=f"{mask_desc_missing.mean()*100:.2f}%",
            severity="Low",
            review_action="Acceptable for numeric aggregation. For product-name reports, optionally backfill from StockCode master map or annotate gap size.",
        ),
    ]

    # Add status column based on thresholds/severity
    for r in rules:
        cnt = r["count"]
        sev = r["severity"].upper()
        if sev == "CRITICAL":
            r["status"] = "FAIL ❌" if cnt > 0 else "PASS ✅"
        elif sev == "HIGH":
            r["status"] = (
                "REVIEW ⚠️" if cnt > 0 else "PASS ✅"
            )  # high-severity findings need attention even if expected
        elif sev == "MEDIUM":
            r["status"] = "NOTED ℹ️" if cnt > 0 else "CLEAN ✨"
        else:  # LOW / REVIEW
            r["status"] = "INFO 💡" if cnt > 0 else "OK ✔️"

    return pd.DataFrame(rules)


# Execute
dq_table = validate_rules(df)

# Optional: reorder columns for cleaner presentation
col_order = [
    "rule_id",
    "category",
    "description",
    "columns",
    "count",
    "pct",
    "severity",
    "status",
    "review_action",
]
dq_table = dq_table[col_order]

print("=== ISSUE LOG ===\n")
display(
    dq_table.style.set_properties(**{"text-align": "left"})
    .format({"count": "{:,}", "pct": "{}"})
    .applymap(
        lambda x: (
            "background-color:#ffebee;color:#b71c1c;font-weight:bold;"
            if "FAIL" in str(x) or "CRITICAL" == x
            else ""
        ),
        subset=["status", "severity"],
    )
    .applymap(
        lambda x: (
            "background-color:#fff3cd;color:#856404;"
            if "REVIEW" in str(x) or "HIGH" == x
            else ""
        ),
        subset=["status", "severity"],
    )
    .applymap(
        lambda x: (
            "background-color:#d1ecf1;color:#0c5460;"
            if "NOTED" in str(x) or "MEDIUM" == x
            else ""
        ),
        subset=["status", "severity"],
    )
    .applymap(
        lambda x: (
            "background-color:#d4edda;color:#155724;"
            if "PASS" in str(x) or "CLEAN" in str(x) or "LOW" == x or "REVIEW" == x
            else ""
        ),
        subset=["status", "severity"],
    )
)

# Save for report
dq_table.to_csv("data_quality_validation_report.csv", index=False)
print("\n💾 Saved to: data_quality_validation_report.csv")
```

## Issue prioritization

```python
# Phase 8 — triage board, keyed to the Phase-7 rule_ids
triage = pd.DataFrame(
    [
        dict(
            rule_id="DQ06",
            impact="Very High",
            actionability="High (format fix + reload)",
            priority=1,
            decision="Repair now — blocking",
        ),
        dict(
            rule_id="DQ04",
            impact="High",
            actionability="Very High (only ~5 rows)",
            priority=2,
            decision="Investigate & correct individually",
        ),
        dict(
            route := dict(
                rule_id="DQ03",
                impact="High (gross vs net)",
                actionability="Route-only (legitimate returns)",
                priority=3,
                decision="Separate into returns view; never mix raw with sales",
            )
        ),
        dict(
            rule_id="DQ01",
            impact="High (customer-scoped only)",
            actionability="Low (unrecoverable)",
            priority=4,
            decision="Exclude from customer analyses + document ~23% anonymity",
        ),
        dict(
            rule_id="DQ05",
            impact="Medium (AOV/margin)",
            actionability="Medium (policy call)",
            priority=4,
            decision="Flag non-revenue; decide include-in-volume / exclude-from-$",
        ),
        dict(
            rule_id="DQ02",
            impact="Medium (independence/count)",
            actionability="High (dedupe)",
            priority=5,
            decision="Inspect samples; drop true repeats, log conflicting",
        ),
        dict(
            rule_id="DQ07",
            impact="Low (readability)",
            actionability="Medium (backfill/annotate)",
            priority=5,
            decision="Backfill from StockCode master or annotate gap",
        ),
    ]
).sort_values("priority")

display(
    triage.style.set_caption(
        "PHASE 8 · ISSUE PRIORITIZATION (Impact × Actionability)"
    ).applymap(
        lambda v: (
            "font-weight:bold;background:#000000;color:#fde2e2" if str(v) == "1" else ""
        ),
        subset=["priority"],
    )
)
triage.to_csv("issue_prioritization_matrix.csv", index=False)
```

## Cleaned Sample

```python
cleaned_df = df.copy()

print(f"Raw Rows: {len(cleaned_df)}")

# 2. APPLY CLEANING RULES 
# Rule DQ06: Parse Dates (Critical)

cleaned_df["InvoiceDate"] = pd.to_datetime(cleaned_df["InvoiceDate"], errors="coerce")

# Rule DQ04 & DQ03: Filter out Negative Price (Anomaly) and Negative Quantity (Returns)

df_clean_sales = cleaned_df[
    (cleaned_df["Quantity"] > 0)
    & (cleaned_df["Price"] >= 0)
    & (cleaned_df["InvoiceDate"].notna())  # Enforce DQ06 pass
].copy()

# Rule DQ02: Handle Duplicates

df_clean_sales = df_clean_sales.drop_duplicates()

# Rule DQ01: CustomerID

# 3. CREATE DERIVED COLUMNS (Optional but recommended for "Cleaned" status)

df_clean_sales["Revenue"] = df_clean_sales["Quantity"] * df_clean_sales["Price"]

df_clean_sales["IsIdentifiedCustomer"] = (
    df_clean_sales["Customer ID"].notna().astype(int)
)


output_sales = "cleaned_online_retail_sales.csv"
df_clean_sales.to_csv(output_sales, index=False)
print(f"✅ Cleaned Sales Sample Saved: {output_sales}")
print(
    f"   Rows: {len(df_clean_sales)} (Removed returns, neg prices, invalid dates, dupes)"
)

# 5. EXPORT THE RETURNS SAMPLE .
df_returns = cleaned_df[
    (cleaned_df["Quantity"] < 0) | (cleaned_df["Invoice"].str.startswith("C", na=False))
].copy()

# Clean returns too (remove dupes, parse dates)
df_returns["InvoiceDate"] = pd.to_datetime(df_returns["InvoiceDate"], errors="coerce")
df_returns = df_returns.drop_duplicates()
df_returns["RevenueImpact"] = (
    df_returns["Quantity"] * df_returns["Price"]
)

output_returns = "cleaned_online_retail_returns.csv"
df_returns.to_csv(output_returns, index=False)
print(f"✅ Cleaned Returns Sample Saved: {output_returns}")
print(f"   Rows: {len(df_returns)}")

# 6. FINAL VALIDATION CHECK 
print("\n--- FINAL CLEANED STATS ---")
print(f"Total Identified Customers: {df_clean_sales['Customer ID'].nunique()}")
print(f"Total Anonymous Transactions: {df_clean_sales['Customer ID'].isna().sum()}")
print(f"Total Revenue (Cleaned): ${df_clean_sales['Revenue'].sum():,.2f}")
print(
    f"Date Range: {df_clean_sales['InvoiceDate'].min()} to {df_clean_sales['InvoiceDate'].max()}"
)
```
