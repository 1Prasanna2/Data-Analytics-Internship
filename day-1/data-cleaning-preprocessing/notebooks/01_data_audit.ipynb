## Loading and Profiling

#### 1.Dataset Shape

```python
import pandas as pd, numpy as np

df = pd.read_csv('../data/raw/SampleSuperstore.csv',
                dtype={'Postal Code': str}, na_values=['-  ','-',' ',''])

df.columns = df.columns.str.strip()

print(f"Shape: {df.shape}")
df.info()
print("\n---- Null Inventory ----")
print(df.isnull().sum())

print(f"\nExact Duplicated Rows: {df.duplicated().sum()}")

print("\n--- Format Sniffing ---")
print(df['Ship Mode'].value_counts(dropna=False))

df['Postal Code'] = df['Postal Code'].apply(lambda x: str(x).zfill(5) if pd.notna(x) else x)

print("\nPostal Code samples:")
print(df['Postal Code'].head(10))

print("\nSales Samples:")
print(df['Sales'].head(10))

print("\nDiscount Dtype: ")
print(df['Discount'].dtype)
```
