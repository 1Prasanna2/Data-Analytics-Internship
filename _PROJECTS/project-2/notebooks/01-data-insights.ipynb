```python

covid = covid[~covid["iso_code"].str.startswith("OWID_", na=False)]

target_countries = [
    "Israel",
    "United Kingdom",
    "United States",
    "Brazil",
    "India",
    "South Africa",
    "Germany",
]
```

```python
covid_subset = covid[covid['location'].isin(target_countries)].copy()

assert not covid_subset.duplicated(
    subset=["iso_code", "date"]
).any(), "CRITICAL ERROR: Duplicate country-date combinations found!"

print(covid_subset["location"].nunique())
print(f"Rows reduced from {len(covid)} to {len(covid_subset)}")
```

```python
ISO = sorted(covid_subset["iso_code"].unique())
ISO
```

```python
ROLL, MIN_PERIODS, TAIL = 7, 3, 14
```

## **SORT**

```python
cs = cs.sort_values(["iso_code", "date"]).reset_index(drop=True)
cutoff = cs["date"].max() - pd.Timedelta(days=TAIL)
cs = cs[cs["date"] <= cutoff].copy()
```

```python
cs = covid_subset.copy()
cs["date"] = pd.to_datetime(cs["date"], format="%Y-%m-%d")
for c in [
    "new_cases",
    "new_deaths",
    "new_vaccinations",
    "new_cases_per_million",
    "new_deaths_per_million",
    "people_fully_vaccinated_per_hundred",
    "population",
]:
    cs[c] = pd.to_numeric(cs[c], errors="coerce")
```

## Re-Index

```python

full = pd.date_range(cs["date"].min(), cutoff, freq="D")
cs = (
    cs.set_index(["iso_code", "date"])
    .reindex(pd.MultiIndex.from_product([ISO, full], names=["iso_code", "date"]))
    .sort_index()
    .reset_index()
)
static = [
    "location",
    "continent",
    "population",
    "median_age",
    "hospital_beds_per_thousand",
]
cs[static] = cs.groupby("iso_code")[static].ffill()
assert (
    cs.groupby("iso_code").size().nunique() == 1
), "reindex failed -> row counts differ"
assert not cs.duplicated(
    ["iso_code", "date"]
).any(), "panel not unique per (country,date)"
```

```python
def roll7(s):
    return s.rolling(ROLL, min_periods=MIN_PERIODS).mean()


for c in [
    "new_cases",
    "new_deaths",
    "new_vaccinations",
    "new_cases_per_million",
    "new_deaths_per_million",
    "people_fully_vaccinated_per_hundred",
]:
    cs[f"{c}_7d"] = cs.groupby("iso_code")[c].transform(roll7)
```

```python
fact_daily = cs.sort_values(["iso_code", "date"]).reset_index(drop=True)
pm = fact_daily.pivot(
    index="date", columns="iso_code", values="new_cases_per_million_7d"
)
pm.plot(figsize=(12, 6), linewidth=2)
plt.title("7‑Day Avg NEW CASES PER MILLION — 7‑Country Comparison (normalized)")
plt.ylabel("New cases per million")
plt.xlabel("Date")
plt.legend(title="iso_code")
plt.tight_layout()
plt.savefig("viz_cases_pm.png", dpi=150)
plt.show()
```

```python
USED = [
    "iso_code",
    "date",
    "new_cases",
    "new_deaths",
    "new_vaccinations",
    "new_cases_per_million",
    "new_deaths_per_million",
    "people_fully_vaccinated_per_hundred",
    "population",
]
null_summary = pd.DataFrame(
    {
        "Column": USED,
        "Null Count": fact_daily[USED].isnull().sum().values,
        "Null %": (fact_daily[USED].isnull().mean() * 100).round(2).values,
    }
)
print(null_summary.to_string(index=False))
```

```python
tidy_long = fact_daily.melt(
    id_vars=["iso_code", "location", "date", "continent"],
    value_vars=[c for c in fact_daily.columns if c.endswith("_7d")],
    var_name="metric",
    value_name="value",
).dropna(subset=["value"])
fact_daily.to_parquet("covid_clean.parquet", index=False)
tidy_long.to_parquet("covid_tidy_long.parquet", index=False)
with open("ASSUMPTIONS.md", "a", encoding="utf-8") as f:
    f.write(
        f"\n## Day 1\n- OWID_* aggregates excluded: 34,124 of 429,435 rows (-> 395,311 real-country rows)\n"
        f"- Countries (iso_code): {', '.join(ISO)}\n"
        f"- ROLL={ROLL}, MIN_PERIODS={MIN_PERIODS}; rolling from new_* / *_per_million (totals never differenced)\n"
        f"- Provisional tail TAIL={TAIL}d dropped; cut date {cutoff.date()} (raw max was 2024-08-14)\n"
        f"- Reindexed to continuous daily calendar so shift()/roll() are day-aligned (not row-aligned)\n"
        f"- Comparisons normalized to per-million / per-100 (brief hint #2)\n"
    )
print(
    f"fact_daily {fact_daily.shape} | tidy_long {tidy_long.shape} | saved + ASSUMPTIONS.md"
)
```
