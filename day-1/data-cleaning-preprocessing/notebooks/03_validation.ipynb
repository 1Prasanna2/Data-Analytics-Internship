## Validating the Cleaned Data File 

```python
import pandas as pd
raw    = pd.read_csv('../data/raw/SampleSuperstore.csv', dtype={'Postal Code': str})
clean  = pd.read_csv('../data/clean/Cleaned_SampleSuperstore.csv', dtype={'Postal Code': str})

def audit(d):
    return {'rows': len(d),
            'missing_cells': int(d.isnull().sum().sum()),
            'duplicate_rows': int(d.duplicated().sum()),
            'zip_valid_%': round(d['Postal Code'].astype(str).str.fullmatch(r'\d{5}').mean()*100, 1)}

summary = pd.DataFrame({'BEFORE': audit(raw), 'AFTER': audit(clean)})
print(summary)  

# Business rules on cleaned data
assert clean['Discount'].between(0, 1).all()
assert (clean['Sales'] >= 0).all()
print("All business rules passed ✅")
```
