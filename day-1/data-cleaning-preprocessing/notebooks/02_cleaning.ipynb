## Cleaning the CSV file

```python
import pandas as pd
import numpy as np

# 1. Load data and handle missing strings
df = pd.read_csv(
    "../data/raw/SampleSuperstore.csv",
    dtype={"Postal Code": str},
    na_values=[" -   ", "-", " ", ""],
)

# 2. Strip whitespace from column names
df.columns = df.columns.str.strip()

# 3. Fix truncated Postal Codes (safe for potential NaNs)
df["Postal Code"] = df["Postal Code"].apply(
    lambda x: str(x).zfill(5) if pd.notna(x) else x
)

# 4. Fill missing Discounts with 0.0
df["Discount"] = df["Discount"].fillna(0.0)

# 5. Drop exact duplicates
df = df.drop_duplicates()

print(f"Final Shape: {df.shape}")
print(f"Total Missing Values:\n{df.isnull().sum()}")
print(f"Total Duplicates Remaining: {df.duplicated().sum()}")
print("\n--- Discount Summary ---")
print(df["Discount"].describe())


df.to_csv('../data/clean/Cleaned_SampleSuperstore.csv', index=False)
```
